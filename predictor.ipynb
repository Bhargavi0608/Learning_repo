```python
import pickle
import pandas as pd
import joblib
```

```python
new_data = pd.read_excel('new_data.xlsx')
new_data.head()
```

```python


# Load it using joblib instead of pickle
trained_model = joblib.load('random_forest_model.pkl')

print(type(trained_model))
with open('model_columns.pkl', 'rb') as f:
    model_columns = pickle.load(f)
```

```python
# 2. Get dummies on your new data
new_df_encoded = pd.get_dummies(new_data , drop_first = True)

# 3. ALIGNMENT STEP: 
# Add any missing columns that were in training but are missing now (fill with 0)
for col in model_columns:
    if col not in new_df_encoded.columns:
        new_df_encoded[col] = 0

# 4. Ensure the exact same column order as training
# (This also drops any brand new columns the model hasn't seen before)
new_df_encoded = new_df_encoded[model_columns]
new_df_encoded.head()

```

```python
predictions = trained_model.predict(new_df_encoded)

print(predictions)
```
