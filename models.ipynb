```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
x = pd.read_csv("train.csv")
df = pd.DataFrame(x)
df['Sex'] = df['Sex'].map({'male': 1, 'female': 0})
df['Embarked'] = df['Embarked'].map({'C' : 1 , 'S' : 2 , 'Q' : 3})
df['Age'] = df['Age'].fillna(df['Age'].mean())
y = df['Survived']
X = df.drop(['Survived' , 'Name', 'Ticket','Cabin'] , axis = 1)

```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.1,
    random_state=42
)
X_train.fillna(X_train['Age'].mean(), inplace = True)

from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import accuracy_score
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt
dtree = DecisionTreeClassifier()
dtree.fit(X_train, y_train)
```

```python
y_pred_DT = dtree.predict(X_test)

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred_DT)
print("Accuracy:", accuracy)
```

```python
# Create logistic regression model using 'liblinear' solver
from sklearn.linear_model import LogisticRegression
logreg = LogisticRegression(solver='liblinear')

# Fit the model to training data
logreg.fit(X_train , y_train)
```

```python
y_pred_LR = logreg.predict(X_test)
```

```python
precision_score(y_test, y_pred_LR)
```

```python
from mlxtend.plotting import plot_decision_regions

plot_decision_regions(X_train, y_train, clf=logreg, legend=2)

#library is not installed. also make sure that there are only 2 features in X
```

```python
from sklearn.tree import DecisionTreeRegressor
regtree = DecisionTreeRegressor(random_state=42)
regtree.fit(X_train, y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier
randomforest = RandomForestClassifier(n_estimators=100, random_state=50)
randomforest.fit(X_train, y_train)
```

```python
y_pred_RF = randomforest.predict(X_test)
accuracy_score(y_pred_RF , y_test)
```

```python
precision_score(y_test, y_pred_RF)
```
